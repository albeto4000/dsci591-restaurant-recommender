```python
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default = 'png'
```

```python
users = pd.read_parquet('data/users.parquet')
with pd.option_context('display.max_columns', None):
    display(users.head())
```

```python
restaurants = pd.read_parquet('data/restaurants.parquet')
with pd.option_context('display.max_columns', None):
    display(restaurants.head())
```

```python
ratings = pd.read_parquet('data/reviews.parquet')

display(ratings.head())
```

```python
n_restaurants = len(restaurants)
n_ratings = len(ratings)
n_users = ratings['user_id'].nunique()

print(f'There are {n_ratings} ratings for {n_restaurants} restaurants, created by {n_users} unique users')
```

```python
density = n_ratings / (n_users * n_restaurants)

print(f'The density of the dataset is {density:.6f}')
```

```python
ratings_per_user = ratings['user_id'].value_counts().reset_index()

fig = px.ecdf(ratings_per_user, x = 'count', template = 'plotly_white', log_x = True)

fig.update_layout(
    xaxis = dict(
    tickmode = 'array',
    tickvals = [0] + [10 ** i for i in range(5)],
    ticktext = ['0'] + ['10<sup>' + str(i) +'</sup>' for i in range(6)],
    ),
    xaxis_title = 'Restaurants Rated',
    yaxis_title = '% of Users'
)

fig.show()
```

```python
ratings_counts = ratings['stars'].value_counts().reset_index()

fig = px.bar(ratings_counts, x = 'stars', y = 'count', template = 'plotly_white')
fig.show()

print(f'Median Rating: {ratings['stars'].median():.4f}')
print(f'Mean Rating: {ratings['stars'].mean():.4f}')
```

```python
res_stats = ratings.groupby('business_id')['user_id'].count().to_frame('n_ratings')

fig = px.ecdf(res_stats, x = 'n_ratings', template = 'plotly_white', log_x = True)

fig.update_layout(
    xaxis = dict(
    tickmode = 'array',
    tickvals = [0] + [10 ** i for i in range(5)],
    ticktext = ['0'] + ['10<sup>' + str(i) +'</sup>' for i in range(6)],
    ),
    xaxis_title = '# of Reviews',
    yaxis_title = '% of Restaurants'
)


fig.show()
```

```python
res_stats = pd.merge(res_stats, restaurants[['business_id', 'name', 'city', 'state', 'attributes', 'categories']], on = 'business_id')

with pd.option_context('display.max_columns', None):
    display(res_stats.nlargest(10, 'n_ratings')[['name', 'city', 'n_ratings', 'attributes', 'categories']])
```

```python
#Determines whether or not a user 'friended' other users
users['has_friends'] = users['friends'].notna()

#Removes users with outlier review counts using the IQR method
rc_iqr = users['review_count'].quantile(.75) - users['review_count'].quantile(.25)
rc_upper = users['review_count'].quantile(.75) + (1.5 * rc_iqr)
rc_lower = users['review_count'].quantile(.25) - (1.5 * rc_iqr)

#Plots the distribution of 
fig = px.box(users[(users['review_count'] >= rc_lower) & (users['review_count'] <= rc_upper)],
            y = 'review_count', color = 'has_friends', template = 'plotly_white',
            color_discrete_sequence=px.colors.qualitative.Antique,
            title = 'Relationship Between Has_Friends Status and Review Count')
fig.update_layout(yaxis_title = '# of Reviews')
fig.show()
```

```python
users['friend_count'] = users['friends'].str.split().str.len().fillna(0)

fc_iqr = users['friend_count'].quantile(.75) - users['friend_count'].quantile(.25)
fc_upper = users['friend_count'].quantile(.75) + (1.5 * fc_iqr)
fc_lower = users['friend_count'].quantile(.25) - (1.5 * fc_iqr)

avg_review_fc = users[(users['friend_count'] >= fc_lower) & (users['friend_count'] <= fc_upper)].groupby('friend_count')['review_count'].mean().reset_index()

avg_review_fc['rolling_avg'] = avg_review_fc.rolling(window = 10, center = True).mean()['review_count']

fig = px.line(avg_review_fc, x = 'friend_count', y = ['review_count', 'rolling_avg'], template = 'plotly_white', 
             title = 'Average Review Count vs Friend Count')

fig.update_layout(xaxis_title = '# of Friends', yaxis_title = 'Average Review Count',
                 legend_title_text = 'Legend')

legend_names = {'review_count': 'Average Review Count', 'rolling_avg': 'Review Count (Rolling Average)'}
fig.for_each_trace(lambda t: t.update(name = legend_names.get(t.name, t.name)))

fig.show()
```
