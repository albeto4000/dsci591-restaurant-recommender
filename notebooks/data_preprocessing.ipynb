```python
import pandas as pd
import json

#Reads the JSON businesses file into a Pandas dataframe
business_df = pd.read_json("../data/yelp_academic_dataset_business.json", lines = True)

#Selects all businesses with category 'Restaurants'
restaurants = business_df[business_df['categories'].str.contains('Restaurants')]

#Writes the restaurants data to a parquet file
restaurants.to_parquet('../data/restaurants.parquet', index = False)
```

```python
#Reads the review JSON file 10,000 rows at a time 
review_df = pd.read_json("../data/yelp_academic_dataset_review.json", lines = True, chunksize=10000)

#Filters for all reviews whose business_id field matches a business whose categories contains 'Restaurants'
reviews = review_df[review_df['business_id'].isin(restaurants['business_id'])]

#Writes the restaurant reviews data to a parquet file
reviews.to_parquet('../data/reviews.parquet', index = False)
```

```python
#Fetches the user IDs of all users who have reviewed at least one restaurant
review_users = reviews['user_id'].unique()

#Reads the user dataset 10,000 rows at a time 
user_df = pd.read_json("../data/yelp_academic_dataset_user.json", lines = True, chunksize=10000)

#Maps the table columns and datatypes to ensure the data is written to parquet correctly
datatypes = {
    'user_id': str,
    'name': str,
    'review_count': int,
    'yelping_since': str,
    'useful': int,
    'funny': int,
    'cool': int,
    'elite': str,
    'friends': str,
    'fans': int,
    'average_stars': float,
    'compliment_hot': int,
    'compliment_more': int
}

#Filters for the subset of users who have reviewed restaurants
users = user_df[user_df['user_id'].isin(review_users)]

#Writes restaurant reviewer data to a parquet file
users.to_parquet('../data/users.parquet', index = False)
```
